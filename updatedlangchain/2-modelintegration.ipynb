### Streaming and Batch

Streaming.........................

Most models can stream their output content while it is being generated. By displaying output progressively, streaming significanty improves user experience, particularly for longer responses. Calling stream() returns an iterator that yields output chunks as they are produced. You can use a look to process eac chunk in real-time.

```python
models.invoke("Write 200 words paragraph in AI")
```

```python
for chunk in models.stream("Write 200 words paragraph on AI"):
    print(chunk.text, end="|", flush=True)
```

Batch

Batching is a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel:

```python
resp = models.batch([
    "Why do AI make thing easy?",
    "How can we make Agentic AI for any issue?",
    "What is Quantum Computing?"
])
for respp in resp:
    print(respp)
```

```python
resp = models.batch([
    "Why do AI make thing easy?",
    "How can we make Agentic AI for any issue?",
    "What is Quantum Computing?"],
config={
    'max_concurrency':5,
})
```
